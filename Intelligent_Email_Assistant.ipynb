```python
%%writefile app.py
from transformers import pipeline
import streamlit as st

# Load model
generator = pipeline("text-generation", model="gpt2")

def generate_reply(email_text):
    prompt = f"Write a professional reply to this email:\n{email_text}"
    response = generator(prompt, max_length=100)
    return response[0]['generated_text']

def categorize_email(email_text):
    email = email_text.lower()

    if "meeting" in email:
        return "Meetings"
    elif "invoice" in email:
        return "Finance"
    elif "project" in email:
        return "Work"
    else:
        return "General"

# UI
st.title(" Intelligent Email Assistant")

email = st.text_area("Enter Email")

if st.button("Generate Reply"):
    reply = generate_reply(email)
    category = categorize_email(email)

    st.subheader("Reply")
    st.write(reply)

    st.subheader("Category")
    st.write(category)
```
